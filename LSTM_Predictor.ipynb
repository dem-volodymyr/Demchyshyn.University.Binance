# LSTM_Predictor

Структурований ноутбук для навчання, оцінки та **візуалізації** LSTM-моделі прогнозу крипто-цін.

## Що робить ноутбук
- Генерує фічі (SMA, RSI, MACD, volatility)
- Навчає LSTM на `Log_Return`
- Конвертує прогноз у простір ціни
- Будує графіки:
  - train/val loss
  - actual vs predicted
  - порівняння метрик по тикерах

# Імпорти, базова ініціалізація, пороги quality gate і списки фіч/тікерів.

```python
import csv
import warnings
from dataclasses import asdict, dataclass
from pathlib import Path

import joblib
import numpy as np
import pandas as pd
import seaborn as sns
import yfinance as yf
import matplotlib.pyplot as plt

from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
from sklearn.preprocessing import MinMaxScaler

from keras import Sequential, ops
from keras.layers import LSTM, Dense, Dropout
from keras.callbacks import EarlyStopping, ReduceLROnPlateau
from keras.optimizers import Adam

warnings.filterwarnings('ignore')

sns.set_theme(style='whitegrid')
pd.set_option('display.max_columns', None)

MODELS_DIR = Path('models')
REPORTS_DIR = Path('reports')
MODELS_DIR.mkdir(exist_ok=True)
REPORTS_DIR.mkdir(exist_ok=True)

FEATURE_COLS = [
    'Close', 'High', 'Low', 'Volume', 'SMA_20', 'SMA_50',
    'RSI_14', 'MACD', 'MACD_Signal', 'Daily_Return', 'Volatility_10',
]

GATE_MIN_R2 = 0.0
GATE_MAX_MAPE = 15.0
GATE_MIN_DIR_ACC = 52.0

TICKERS = [
    'BTC-USD', 'ETH-USD', 'BNB-USD', 'SOL-USD', 'XRP-USD',
    'ADA-USD', 'DOGE-USD', 'DOT-USD', 'MATIC-USD', 'LTC-USD',
]
```

# directional loss, quality gate, метрики, best-effort збереження model/scaler/meta.

```python
def directional_huber_loss(y_true, y_pred):
    """Huber + penalty for wrong direction to improve Directional Accuracy."""
    err = y_true - y_pred
    abs_err = ops.abs(err)
    huber = ops.where(abs_err <= 1.0, 0.5 * err**2, abs_err - 0.5)
    huber = ops.mean(huber)
    wrong = ops.cast(ops.sign(y_true) != ops.sign(y_pred), 'float32')
    return huber + 0.4 * ops.mean(wrong)


@dataclass
class TrainMetrics:
    mae: float
    rmse: float
    mape: float
    r2: float
    directional_acc: float

    def passes_gate(self) -> bool:
        return (
            self.r2 >= GATE_MIN_R2
            and self.mape <= GATE_MAX_MAPE
            and self.directional_acc >= GATE_MIN_DIR_ACC
        )


@dataclass
class TrainConfig:
    ticker: str = 'BTC-USD'
    start_date: str = '2019-01-01'
    test_ratio: float = 0.15
    lstm1: int = 48
    lstm2: int = 24
    dropout: float = 0.25
    lr: float = 5e-4
    epochs: int = 40
    batch_size: int = 64
    window_size: int = 60
    loss_name: str = 'huber'  # huber | directional


class CryptoPredictor:
    def __init__(self, cfg: TrainConfig):
        self.cfg = cfg
        self.end_date = pd.Timestamp.today().strftime('%Y-%m-%d')
        self.scaler = MinMaxScaler(feature_range=(0, 1))
        self.model = None
        self.history = None
        self.df_raw = None
        self.df_features = None
        self.y_mean = None
        self.y_std = None
        self.prev_close_test = None
        self.X_val = None
        self.y_val = None

    def download_data(self):
        df = yf.download(
            self.cfg.ticker,
            start=self.cfg.start_date,
            end=self.end_date,
            auto_adjust=True,
            progress=False,
        )
        if isinstance(df.columns, pd.MultiIndex):
            df.columns = df.columns.get_level_values(0)
        self.df_raw = df.dropna()
        return self.df_raw

    @staticmethod
    def _calc_rsi(series, period=14):
        delta = series.diff()
        gain = delta.clip(lower=0)
        loss = -delta.clip(upper=0)
        avg_gain = gain.ewm(com=period - 1, min_periods=period).mean()
        avg_loss = loss.ewm(com=period - 1, min_periods=period).mean()
        rs = avg_gain / (avg_loss + 1e-10)
        return 100 - (100 / (1 + rs))

    @staticmethod
    def _calc_macd(series):
        ema_fast = series.ewm(span=12, adjust=False).mean()
        ema_slow = series.ewm(span=26, adjust=False).mean()
        macd_line = ema_fast - ema_slow
        signal_line = macd_line.ewm(span=9, adjust=False).mean()
        return macd_line, signal_line

    def engineer_features(self):
        if self.df_raw is None:
            self.download_data()

        df = self.df_raw.copy()
        close = df['Close']
        df['SMA_20'] = close.rolling(window=20).mean()
        df['SMA_50'] = close.rolling(window=50).mean()
        df['RSI_14'] = self._calc_rsi(close)
        df['MACD'], df['MACD_Signal'] = self._calc_macd(close)
        df['Daily_Return'] = close.pct_change()
        df['Volatility_10'] = df['Daily_Return'].rolling(window=10).std()
        df['Log_Return'] = np.log(close / close.shift(1))

        self.df_features = df[FEATURE_COLS + ['Log_Return']].dropna()
        return self.df_features

    def prepare_data(self):
        features = self.df_features[FEATURE_COLS].values
        target = self.df_features['Log_Return'].values
        closes = self.df_features['Close'].values

        n_test = int(len(features) * self.cfg.test_ratio)
        n_train = len(features) - n_test

        self.scaler.fit(features[:n_train])
        scaled_features = self.scaler.transform(features)

        self.y_mean = float(target[:n_train].mean())
        self.y_std = float(target[:n_train].std() + 1e-8)
        target_scaled = (target - self.y_mean) / self.y_std

        X, y, prev_close = [], [], []
        for i in range(self.cfg.window_size, len(scaled_features)):
            X.append(scaled_features[i - self.cfg.window_size : i])
            y.append(target_scaled[i])
            prev_close.append(closes[i - 1])

        X = np.array(X)
        y = np.array(y)
        prev_close = np.array(prev_close)

        split = n_train - self.cfg.window_size
        self.prev_close_test = prev_close[split:]

        val_split = int(len(X[:split]) * 0.9)
        self.X_val = X[val_split:split]
        self.y_val = y[val_split:split]

        return X[:split], y[:split], X[split:], y[split:]

    def build_model(self, n_features):
        model = Sequential([
            LSTM(self.cfg.lstm1, return_sequences=True, input_shape=(self.cfg.window_size, n_features)),
            Dropout(self.cfg.dropout),
            LSTM(self.cfg.lstm2),
            Dropout(self.cfg.dropout),
            Dense(16, activation='relu'),
            Dense(1),
        ])
        loss = directional_huber_loss if self.cfg.loss_name == 'directional' else 'huber'
        model.compile(optimizer=Adam(self.cfg.lr), loss=loss)
        self.model = model

    def train(self, X_train, y_train):
        callbacks = [
            EarlyStopping(monitor='val_loss', patience=8, restore_best_weights=True),
            ReduceLROnPlateau(monitor='val_loss', factor=0.5, patience=4, min_lr=1e-6),
        ]
        self.history = self.model.fit(
            X_train,
            y_train,
            epochs=self.cfg.epochs,
            batch_size=self.cfg.batch_size,
            validation_data=(self.X_val, self.y_val),
            callbacks=callbacks,
            verbose=0,
        )
        return self.history

    def evaluate_model(self, X_test, y_test):
        pred_scaled = self.model.predict(X_test, verbose=0).flatten()
        pred_log_ret = pred_scaled * self.y_std + self.y_mean
        true_log_ret = y_test.flatten() * self.y_std + self.y_mean

        preds = self.prev_close_test * np.exp(pred_log_ret)
        trues = self.prev_close_test * np.exp(true_log_ret)

        mae = float(mean_absolute_error(trues, preds))
        rmse = float(np.sqrt(mean_squared_error(trues, preds)))
        mape = float(np.mean(np.abs((trues - preds) / (trues + 1e-8))) * 100)
        r2 = float(r2_score(trues, preds))
        dir_acc = float(np.mean(np.sign(pred_log_ret) == np.sign(true_log_ret)) * 100)

        metrics = TrainMetrics(mae=mae, rmse=rmse, mape=mape, r2=r2, directional_acc=dir_acc)
        return trues, preds, true_log_ret, pred_log_ret, metrics

    def _signal_bias(self):
        if self.X_val is None or len(self.X_val) == 0:
            return 0.0
        pred_scaled = self.model.predict(self.X_val, verbose=0).flatten()
        val_preds = pred_scaled * self.y_std + self.y_mean
        return float(np.mean(val_preds))

    def save(self, metrics: TrainMetrics):
        name = self.cfg.ticker.replace('-', '_').lower()
        self.model.save(MODELS_DIR / f'{name}_model.keras')
        joblib.dump(self.scaler, MODELS_DIR / f'{name}_scaler.gz')

        meta = {
            'version': 2,
            'target': 'log_return',
            'feature_cols': FEATURE_COLS,
            'y_mean': float(self.y_mean),
            'y_std': float(self.y_std),
            'sequence_length': int(self.cfg.window_size),
            'quality': {
                'mape': metrics.mape,
                'r2': metrics.r2,
                'directional_acc': metrics.directional_acc,
            },
            'quality_pass': metrics.passes_gate(),
            'signal_bias': self._signal_bias(),
        }
        joblib.dump(meta, MODELS_DIR / f'{name}_meta.gz')
```

```python
# Допоміжні функції: візуалізація, багато-спробний train-until-gate,запис/merge CSV та структура повного звіту.
```

```python
def plot_learning_curves(history, title='Learning Curves'):
    hist = pd.DataFrame(history.history)
    plt.figure(figsize=(8, 4))
    plt.plot(hist['loss'], label='train_loss')
    plt.plot(hist['val_loss'], label='val_loss')
    plt.title(title)
    plt.xlabel('Epoch')
    plt.ylabel('Loss')
    plt.legend()
    plt.tight_layout()
    plt.show()


def plot_actual_vs_pred(trues, preds, ticker='BTC-USD', max_points=250):
    n = min(len(trues), max_points)
    plt.figure(figsize=(11, 4))
    plt.plot(trues[-n:], label='Actual')
    plt.plot(preds[-n:], label='Predicted')
    plt.title(f'{ticker}: Actual vs Predicted (last {n} points)')
    plt.xlabel('Test index')
    plt.ylabel('Price (USD)')
    plt.legend()
    plt.tight_layout()
    plt.show()


def candidate_configs(ticker: str):
    return [
        TrainConfig(ticker=ticker, lstm1=48, lstm2=24, dropout=0.25, lr=5e-4, loss_name='huber'),
        TrainConfig(ticker=ticker, lstm1=64, lstm2=32, dropout=0.2, lr=1e-3, loss_name='directional'),
        TrainConfig(ticker=ticker, lstm1=32, lstm2=16, dropout=0.35, lr=3e-4, window_size=45, loss_name='directional'),
        TrainConfig(ticker=ticker, lstm1=56, lstm2=28, dropout=0.3, lr=4e-4, window_size=75, loss_name='huber'),
        TrainConfig(ticker=ticker, lstm1=40, lstm2=20, dropout=0.2, lr=2e-4, epochs=55, loss_name='directional'),
        TrainConfig(ticker=ticker, lstm1=72, lstm2=36, dropout=0.15, lr=8e-4, epochs=45, loss_name='directional'),
    ]


def write_metrics_csv(results, merge=False):
    path = REPORTS_DIR / 'metrics_extended.csv'
    if merge and path.exists():
        old = pd.read_csv(path)
        new = pd.DataFrame(results)
        merged = pd.concat([old[~old['Криптовалюта'].isin(new['Криптовалюта'])], new])
        merged.to_csv(path, index=False)
        print(f'Metrics merged: {path}')
        return

    with path.open('w', newline='', encoding='utf-8') as fh:
        writer = csv.DictWriter(
            fh,
            fieldnames=[
                'Криптовалюта', 'MAE ($)', 'RMSE ($)', 'MAPE (%)',
                'sMAPE (%)', 'R2', 'Directional Acc (%)', 'Gate Pass',
                'Best Config',
            ],
        )
        writer.writeheader()
        for row in results:
            writer.writerow(row)
    print(f'Metrics saved: {path}')


def train_ticker_until_pass(ticker):
    best = None
    best_cfg = None
    best_predictor = None
    best_artifacts = None

    for i, cfg in enumerate(candidate_configs(ticker), start=1):
        print(f'  [{ticker}] attempt {i}/6 ...')
        predictor = CryptoPredictor(cfg)
        predictor.download_data()

        if predictor.df_raw is None or len(predictor.df_raw) < 400:
            print(f'  [{ticker}] insufficient data')
            continue

        predictor.engineer_features()
        X_train, y_train, X_test, y_test = predictor.prepare_data()
        predictor.build_model(X_train.shape[2])
        history = predictor.train(X_train, y_train)
        trues, preds, true_lr, pred_lr, metrics = predictor.evaluate_model(X_test, y_test)

        print(
            f'  [{ticker}] R2={metrics.r2:.3f} MAPE={metrics.mape:.2f}% '
            f'DirAcc={metrics.directional_acc:.1f}%'
        )

        if best is None or metrics.r2 > best.r2:
            best = metrics
            best_cfg = cfg
            best_predictor = predictor
            best_artifacts = (history, trues, preds, true_lr, pred_lr)

        if metrics.passes_gate():
            predictor.save(metrics)
            return True, metrics, cfg, predictor, (history, trues, preds, true_lr, pred_lr)

    if best_predictor is not None and best is not None:
        best_predictor.save(best)

    return False, best, best_cfg, best_predictor, best_artifacts
```

# Навчання моделі
1) для кожного з 10 тікерів робить до 6 спроб різних конфігів;
2) зберігає PASS-модель або best-effort;
3) формує metrics_df + CSV;
4) показує підсумок gate-pass.

```python
print('=' * 60)
print('Training until gate passes')
print(f'Gate: R2>={GATE_MIN_R2}, MAPE<={GATE_MAX_MAPE}%, DirAcc>={GATE_MIN_DIR_ACC}%')
print('=' * 60)

rows = []
last_ticker_artifacts = None
passed = []

for ticker in TICKERS:
    print(f'\n### {ticker}')
    try:
        ok, metrics, cfg, predictor, artifacts = train_ticker_until_pass(ticker)
        if metrics is None or cfg is None:
            print(f'  [{ticker}] skipped (no metrics)')
            continue

        row = {
            'Криптовалюта': ticker,
            'MAE ($)': metrics.mae,
            'RMSE ($)': metrics.rmse,
            'MAPE (%)': metrics.mape,
            'sMAPE (%)': metrics.mape,
            'R2': metrics.r2,
            'Directional Acc (%)': metrics.directional_acc,
            'Gate Pass': 'YES' if ok else 'NO',
            'Best Config': str(asdict(cfg)),
        }
        rows.append(row)

        if artifacts is not None:
            history, trues, preds, true_lr, pred_lr = artifacts
            last_ticker_artifacts = (ticker, history, trues, preds)

        if ok:
            passed.append(ticker)
            print('  PASS — saved to models/')
        else:
            print('  FAIL — saved best attempt (below gate)')

    except Exception as exc:
        print(f'  [{ticker}] failed: {exc}')

write_metrics_csv(rows, merge=False)
metrics_df = pd.DataFrame(rows).sort_values('R2', ascending=False)

print('\n' + '=' * 60)
print(f'Passed gate: {len(passed)}/{len(TICKERS)}')
for t in passed:
    print(f'  - {t}')

metrics_df
```

# Додаткова візуалізація для ОСТАННЬОГО успішно натренованого тікера із batch-пайплайна (learning curves + actual/predicted).

```python
if last_ticker_artifacts is not None:
    t, history_t, trues_t, preds_t = last_ticker_artifacts
    plot_learning_curves(history_t, title=f'{t} learning curves')
    plot_actual_vs_pred(trues_t, preds_t, ticker=t)
else:
    print('Немає артефактів для візуалізації (усі тікери впали).')
```
